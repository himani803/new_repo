```python
import numpy as np
from sklearn.preprocessing import PowerTransformer
```

```python
#A right-skewed feature
X=np.array([ [-1],[0],[3],[5],[10],[50] ])
```

```python
#create the transformer
pt= PowerTransformer(method="yeo-johnson")
```

```python
#fit and transform
X_transformed= pt.fit_transform(X)
print("Original:")
print(X.ravel())
print("\nTransformed:")
print(np.round(X_transformed.ravel(),3))
print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
#A right-skewed feature
Y=np.array([ [1],[1],[3],[5],[10],[50] ])

#create the transformer
pt= PowerTransformer(method="box-cox")

#fit and transform
Y_transformed= pt.fit_transform(Y)
print("Original:")
print(X.ravel())
print("\nTransformed:")
print(np.round(X_transformed.ravel(),3))
print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
import numpy as np
import pandas as pd

df2 = pd.DataFrame({
    # Numerical columns
    'age': [25.0, 30.0, np.nan, 35.0, 42.0, np.nan, 28.0],
    'salary': [50000.0, np.nan, 65000.0, 72000.0, np.nan, 95000.0, 58000.0],
    'experience': [1.0, 3.0, 5.0, np.nan, 12.0, 15.0, 2.0],

    # Categorical columns
    'education': ['High School', 'Bachelors', np.nan, 'Masters', 'PhD', 'Bachelors', np.nan],  # Ordinal
    'department': ['IT', 'HR', 'Finance', np.nan, 'IT', 'Finance', 'HR'],                       # Nominal
    'city': ['Mumbai', 'Delhi', np.nan, 'Bangalore', 'Mumbai', np.nan, 'Delhi']                 # Nominal
})

print(df2)

```

```python
df2
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer= SimpleImputer(strategy='mean')
imputer.fit_transform(df2[['salary']])
```

```python
imputer=SimpleImputer(strategy='mean')
df2['new_salary']= imputer.fit_transform(df2[['salary']]).flatten()
```

```python
imputer=SimpleImputer(strategy='most_frequent')
df2['city']= imputer.fit_transform(df2[['city']]).flatten()
```
