 Feature Scaling and Distance Calculation

```python

import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = {
    'Age': [22, 25, 30, 35, 28, 40],
    'Income': [200000, 250000, 800000, 900000, 400000, 1000000],
    'Class': ['Buy', 'Buy', 'Not Buy', 'Not Buy', 'Buy', 'Not Buy']
}

df = pd.DataFrame(data)

X = df[['Age', 'Income']]
y = df['Class']


new = pd.DataFrame([[29, 450000]], columns=['Age', 'Income'])


model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

print("Without Scaling:", model.predict(new))


scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)
new_scaled = scaler.transform(new)

model.fit(X_scaled, y)

print("Min-Max:", model.predict(new_scaled))


scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
new_scaled = scaler.transform(new)

model.fit(X_scaled, y)

print("Standardization:", model.predict(new_scaled))
```
