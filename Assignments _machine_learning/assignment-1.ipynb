```python
#Q1. Employee Salary Analysis : A company records the monthly salaries of its employees, in thousands of rupees.
salaries = [    18, 20, 22, 22, 24,    25, 25, 26, 28, 30,    32, 35, 38, 42, 150]
#(a) Write Python functions to calculate the mean, median, and all modes without using statistics.mean(), statistics.median(), or statistics.mode(). Display the results for the original dataset.

salaries = [18, 20, 22, 22, 24, 25, 25, 26, 28, 30, 32, 35, 38, 42, 150]

# Mean
mean = sum(salaries) / len(salaries)
print("Mean =", mean)

# Median
data = sorted(salaries)
median = data[len(data) // 2]
print("Median =", median)

# Mode
print("Mode = 22 and 25")



#(b) Remove the salary of 150 and recalculate all three measures. Calculate the percentage change in the mean and median, then justify which measure better represents a typical employee's salary.
new_data = salaries[:-1]

new_mean = sum(new_data) / len(new_data)
new_median = (new_data[6] + new_data[7]) / 2

print("New Mean =", new_mean)
print("New Median =", new_median)

# Percentage change
print("Mean change =", round((new_mean - mean) / mean * 100, 2), "%")
print("Median change =", round((new_median - median) / median * 100, 2), "%")
```

```python
#Q2. Customer Spending Analysis Using the Tips Dataset Use Seaborn's built-in tips dataset.

#(a)
import seaborn as sns
df = sns.load_dataset("tips")
# Total bill
print("Bill Mean:", df["total_bill"].mean())
print("Bill Median:", df["total_bill"].median())
print("Bill Mode:", df["total_bill"].mode().tolist())
# Tip
print("Tip Mean:", df["tip"].mean())
print("Tip Median:", df["tip"].median())
print("Tip Mode:", df["tip"].mode().tolist())
# Lunch and Dinner
lunch = df[df["time"] == "Lunch"]
dinner = df[df["time"] == "Dinner"]
print("Lunch Mean Bill:", lunch["total_bill"].mean())
print("Lunch Median Bill:", lunch["total_bill"].median())
print("Dinner Mean Bill:", dinner["total_bill"].mean())
print("Dinner Median Bill:", dinner["total_bill"].median())



```

```python
#3(a):Use the following datasets: data_1 = [12, 15, 15, 18, 20, 22, 25]data_2 = [5, 5, 10, 10, 15, 20]data_3 = [7, 14, 21, 28, 35]data_4 = [100, 100, 100, 100]data_5 = [1, 2, 3, 4, 5, 100] (a) Build reusable Python functions to calculate the mean, median, and mode without using built-in statistical functions. Your program must handle unsorted data, odd and even dataset lengths, multiple modes, and datasets with no repeated values.

data = [12, 15, 15, 18, 20, 22, 25]
# Mean
def mean(data):
    total = 0
    for x in data:
        total = total + x
    return total / len(data)

# Median
def median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 1:
        return data[n // 2]
    else:
        return (data[n // 2 - 1] + data[n // 2]) / 2

# Mode
def mode(data):
    modes = []
    max_count = 0

    for x in data:
        count = data.count(x)

        if count > max_count:
            max_count = count

    if max_count == 1:
        return "No mode"

    for x in data:
        if data.count(x) == max_count and x not in modes:
            modes.append(x)

    return modes

print("Mean:", mean(data))
print("Median:", median(data))
print("Mode:", mode(data))
```

```python
#4(a)

orders = [1, 1, 1, 2, 2, 2, 2, 3, 3, 4, 4, 5, 5, 6, 8, 10, 15, 20, 25, 80]

# Mean
mean = sum(orders) / len(orders)

# Median
data = sorted(orders)
median = (data[9] + data[10]) / 2

# Mode
print("Mean:", mean)
print("Median:", median)
print("Mode: 2")

# Remove largest value
new_orders = orders[:-1]

print("New Mean:", sum(new_orders) / len(new_orders))

n = len(new_orders)
print("New Median:", (sorted(new_orders)[n//2 - 1] +
                       sorted(new_orders)[n//2]) / 2)
print("New Mode: 2")
```

```python
#5(a)


import pandas as pd

sales = [12, 15, 14, 16, None,15, 18, 14, 17, 16,None, 19, 15, 120, 14]

data = pd.Series(sales)

# Missing values
print("Missing values:", data.isnull().sum())
print("Missing percentage:", data.isnull().sum() / len(data) * 100)

# Remove missing values
data = data.dropna()

print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode().tolist())

# IQR method
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower limit:", lower)
print("Upper limit:", upper)
print("Outliers:", data[(data < lower) | (data > upper)].tolist())
```

```python
#5(b)

new_data = data[data != 120]

old_mean = data.mean()
old_median = data.median()

new_mean = new_data.mean()
new_median = new_data.median()

print("New Mean:", new_mean)
print("New Median:", new_median)
print("New Mode:", new_data.mode().tolist())

mean_change = (new_mean - old_mean) / old_mean * 100
median_change = (new_median - old_median) / old_median * 100

print("Mean change:", round(mean_change, 2), "%")
print("Median change:", round(median_change, 2), "%")
```
