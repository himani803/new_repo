Question 1: KNN Mathematical and Distance Calculation Challenge

```python
import pandas as pd 

data = {
    'Age': [22, 25, 28, 35, 30, 40, 24, 38],
    'Income': [3, 4, 5, 8, 6, 10, 7, 9],
    'Class': ['Purchased', 'Purchased', 'Not Purchased',
              'Not Purchased', 'Purchased', 'Not Purchased',
              'Purchased', 'Not Purchased']
}

df=pd.DataFrame(data)
print(df)

X = df[['Age', 'Income']]
y = df['Class']
new_customer = [[29, 6]]

new_customer = pd.DataFrame([[29, 6]], columns=['Age', 'Income'])

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)

knn.fit(X, y)

result = knn.predict(new_customer)

print(result)
```

Question 2: Small K vs. Large K and Bias-Variance Analysis

```python
k =[1,3,5,9,15,25]

train = [100, 96, 93, 88, 80, 72]

test = [72, 82, 87, 85, 78, 70]

for i in range(len(k)):
    gap = train[i] - test[i]
    print(gap)

    gap = 100 - 72

    gap = 96 - 82

    best = test.index(max(test))

    print(k[best])
```
