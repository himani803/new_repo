# Write Python functions to calculate the mean, median, and all modes without using statistics.mean(), statistics.median(), or statistics.mode(). Display the results for the original dataset.

```python
salaries = [    18, 20, 22, 22, 24,    25, 25, 26, 28, 30,    32, 35, 38, 42, 150]
```

```python
import statistics
```

```python
# 1. MEAN FUNCTION 
def get_mean(data):
    total = sum(data)      
    count = len(data)      
    return total / count   

# 2. MEDIAN FUNCTION 
def get_median(data):
    data.sort()            
    count = len(data)
    middle = count // 2    
    
    if count % 2 != 0:     
        return data[middle]
    else:                 
        return (data[middle - 1] + data[middle]) / 2

# 3. MODE FUNCTION 
def get_mode(data):
    highest_count = 0
    
    for num in data:
        if data.count(num) > highest_count:
            highest_count = data.count(num)
            
   
    modes = []
    for num in data:
        if data.count(num) == highest_count and num not in modes:
            modes.append(num)
            
    return modes




print("Mean:", get_mean(salaries))
print("Median:", get_median(salaries))
print("Mode:", get_mode(salaries))
```

# Remove the salary of 150 and recalculate all three measures. Calculate the percentage change in the mean and median, then justify which measure better represents a typical employee's salary.

```python

old_salaries = [18, 20, 22, 22, 24, 25, 25, 26, 28, 30, 32, 35, 38, 42, 150]
old_mean = sum(old_salaries) / 15
old_median = old_salaries[7]  


new_salaries = [18, 20, 22, 22, 24, 25, 25, 26, 28, 30, 32, 35, 38, 42]
new_mean = sum(new_salaries) / 14
new_median = (new_salaries[6] + new_salaries[7]) / 2  


mean_drop = ((old_mean - new_mean) / old_mean) * 100
median_drop = ((old_median - new_median) / old_median) * 100

print("Old Mean:", old_mean, " Old Median:", old_median)
print("New Mean:", round(new_mean, 2), " New Median:", new_median)
print("Mean Drop:", round(mean_drop, 1), "%")
print("Median Drop:", round(median_drop, 1), "%")
```

Use Seaborn's built-in tips dataset.

(a) Calculate the mean, median, and all modes of total_bill and tip. Then divide the dataset into Lunch and Dinner groups and calculate the mean and median bill for each group.

```python
import seaborn as sns
tips = sns.load_dataset("tips")

old_mean = tips['total_bill'].mean()
old_median = tips['total_bill'].median()

new_tips = tips.sort_values('total_bill').iloc[:-3]
new_mean = new_tips['total_bill'].mean()
new_median = new_tips['total_bill'].median()

drop_mean = ((old_mean - new_mean) / old_mean) * 100
drop_median = ((old_median - new_median) / old_median) * 100

print("Old Mean:", old_mean, " New Mean:", new_mean, " Drop:", round(drop_mean, 2), "%")
print("Old Median:", old_median, " New Median:", new_median, " Drop:", round(drop_median, 2), "%")
```

(a) Build reusable Python functions to calculate the mean, median, and mode without using built-in statistical functions. Your program must handle unsorted data, odd and even dataset lengths, multiple modes, and datasets with no repeated values.

(b) Run your calculator on all five datasets and generate a formatted report. Identify which dataset is most affected by an extreme value and demonstrate numerically why its mean and median behave differently

```python
data_1 = [12, 15, 15, 18, 20, 22, 25] 
data_2 = [5, 5, 10, 10, 15, 20]
data_3 = [7, 14, 21, 28, 35]
data_4 = [100, 100, 100, 100] 
data_5 = [1, 2, 3, 4, 5, 100]
```

```python
def get_mean(data):
    return sum(data) / len(data)

def get_median(data):
    data.sort()
    mid = len(data) // 2
    if len(data) % 2 == 0:
        return (data[mid - 1] + data[mid]) / 2 
    else:
        return data[mid]                       

def get_mode(data):
    highest = max([data.count(num) for num in data])
    if highest == 1:
        return "No Mode"

    modes = []
    for num in data:
        if data.count(num) == highest and num not in modes:
            modes.append(num)
    return modes        
print("Data 1  Mean:", round(get_mean(data_1), 2), " Median:", get_median(data_1), " Mode:", get_mode(data_1))
print("Data 2  Mean:", round(get_mean(data_2), 2), " Median:", get_median(data_2), " Mode:", get_mode(data_2))
print("Data 3 Mean:", round(get_mean(data_3), 2), "Median:", get_median(data_3), " Mode:", get_mode(data_3))
print("Data 4  Mean:", round(get_mean(data_4), 2), " Median:", get_median(data_4), " Mode:", get_mode(data_4))
print("Data 5  Mean:", round(get_mean(data_5), 2), " Median:", get_median(data_5), " Mode:", get_mode(data_5))
    
```
