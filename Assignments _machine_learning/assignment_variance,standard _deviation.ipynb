Employee Performance: Population vs. Sample Variance

```python
import numpy as np 
import pandas as pd
```

```python
scores = [45, 50, 52, 55, 58, 60, 65, 75]
data = np.array(scores)

N = len(data)
mean = sum(data)/N
squared_diff= (data - mean )**2
pop_var= sum(squared_diff)/N
sample_var= sum(squared_diff)/(N-1)
pop_std = pop_var ** 0.5
sample_std = sample_var ** 0.5
print("Mean:", mean)
print("Population Variance:", pop_var)
print("Sample Variance:", sample_var)
print("Population Standard Deviation:", pop_std)
print("Sample Standard Deviation:", sample_std)



```

Customer Spending: Comparing Data Spread
Use Seaborn's built-in tips dataset. Compare the total_bill and tip columns.

```python
import seaborn as sns

tips = sns.load_dataset("tips")

columns = ["total_bill", "tip"]


def get_stats(data):
    result = []

    for col in columns:

        values = data[col].to_numpy()
        n = len(values)

        # Mean
        mean = np.sum(values) / n

        # Population variance
        variance = np.sum((values - mean) ** 2) / n

        # Standard deviation
        std = np.sqrt(variance)

        # Coefficient of variation
        cv = (std / mean) * 100

        result.append([col, mean, variance, std, cv])

    return pd.DataFrame(
        result,
        columns=["Column", "Mean", "Variance", "Std Dev", "CV (%)"]
    ).set_index("Column")



before = get_stats(tips)

print(before.round(4))
```

Build a Variance and Standard Deviation Calculator
Create a reusable Python program for the following datasets.

```python

def mean(data):
    return sum(data) / len(data)


def variance(data):
    avg = mean(data)
    total = 0

    for x in data:
        total += (x - avg) ** 2

    return total / len(data)



def sample_variance(data):
    avg = mean(data)
    total = 0

    for x in data:
        total += (x - avg) ** 2

    return total / (len(data) - 1)



def std(data):
    return variance(data) ** 0.5



def sample_std(data):
    return sample_variance(data) ** 0.5


# Datasets
data_1 = [10, 20, 30, 40, 50]
data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]



datasets = [data_1, data_2, data_3, data_4]

for data in datasets:
    print("\nData:", data)
    print("Mean:", mean(data))
    print("Population Variance:", variance(data))
    print("Sample Variance:", sample_variance(data))
    print("Population SD:", std(data))
    print("Sample SD:", sample_std(data))
```

 Investigating the Effect of Mathematical Transformations

```python

measurements = [12, 15, 18, 20, 25, 30, 40]


add_data = [x + 10 for x in measurements]
multiply_data = [x * 4 for x in measurements]

# Mean
def mean(data):
    return sum(data) / len(data)

# Population Variance
def variance(data):
    avg = mean(data)
    return sum((x - avg)**2 for x in data) / len(data)

# Standard Deviation
def std(data):
    return variance(data)**0.5

datasets = {
    "Original": measurements,
    "Add 10": add_data,
    "Multiply by 4": multiply_data
}

for name, data in datasets.items():
    print("\n", name)
    print("Mean:", round(mean(data), 2))
    print("Variance:", round(variance(data), 2))
    print("Standard Deviation:", round(std(data), 2))
```

 Outlier Detection and Robust Spread Analysis

```python

import statistics

sales = [12, 14, 15, 15, 16, 17, 18,
         19, 20, 21, 22, 24, 25, 120]


mean = statistics.mean(sales)
median = statistics.median(sales)


variance = statistics.pvariance(sales)
std = statistics.pstdev(sales)

q1 = statistics.median(sales[:7])
q3 = statistics.median(sales[7:])

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = []

for x in sales:
    if x < lower or x > upper:
        outliers.append(x)

print("Mean:", round(mean, 2))
print("Median:", median)
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(std, 2))
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Outliers:", outliers)
```
