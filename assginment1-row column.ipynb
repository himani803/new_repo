```python
#Q1. Student Performance DataFrame: Construction and Analysis A college wants to analyze student performance using the following data.

data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Vikas", "Pooja","Karan", "Simran", "Arjun", "Meena", "Rohit", "Anjali"],
    "Age": [19, 20, 18, 21, 20, 19, 22, 18, 21, 20, 19, 22],
    "Hours_Studied": [2, 5, 3, 8, 6, 4, 9, 2, 7, 5, 6, 8],
    "Attendance": [65, 85, 70, 95, 88, 75, 98, 60, 90, 82, 86, 94],
    "Marks": [48, 78, 55, 92, 84, 67, 96, 45, 89, 76, 81, 93]
}

df = pd.DataFrame(data)

# First five records
print(df.head())

# Last three records
print(df.tail(3))

# Rows and columns
print("Shape:", df.shape)

# Column names
print("Columns:", df.columns)

# Select some columns
print(df[["Name", "Hours_Studied", "Marks"]])

# Find average marks
avg = df["Marks"].mean()
print("Average marks:", avg)

# Students who scored above average
result = df[df["Marks"] > avg]
print(result[["Name", "Hours_Studied", "Marks"]])

# Reset index
result = result.reset_index(drop=True)
print("After resetting index:")
print(result)
```

```python
#2.Using the DataFrame created in Q1, perform the following operations.

# Select rows using iloc
print("Question 1:")
print(df.iloc[2:8][["Name", "Attendance", "Marks"]])
# iloc me last position include nahi hoti.
#  Set Student_ID as index
print("Question 2:")
df2 = df.copy()
df2.index = df2["Student_ID"]
print(df2.loc[[103, 107, 110]])

#Students who meet all conditions
print("Question 3:")
print(df[
    (df["Hours_Studied"] >= 6) &
    (df["Attendance"] >= 85) &
    (df["Marks"] > 80)
])

#Last four rows and last three columns
print("Question 4:")
print(df.iloc[-4:, -3:])

# Students scoring below median
print("Question 5:")
median = df["Marks"].median()
result = df[df["Marks"] < median]
print(result)

# Reset index
result = result.reset_index(drop=True)
print(result)
```
